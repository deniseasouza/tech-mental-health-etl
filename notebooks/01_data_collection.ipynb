```python
# Import libraries

import requests
import json
import pandas as pd
```

```python
# Set the URL after running api_fase01.py

url = r'http://127.0.0.1:5000/country'
```

```python
# Import data from API Country

country_data_api = requests.get(url)
```

```python
# Analyze status and type of data

print(country_data_api.status_code)
print(type(country_data_api.text[:200]))
print(country_data_api.encoding)
```

```python
# Convert data api in Dict

country_data = json.loads(country_data_api.text)
type(country_data)
```

```python
# Convert dictionary country_data in DataFrame

df_country = pd.DataFrame.from_dict(country_data, orient='Index')
df_country.head()
```

```python
#Manipulation DataFrame

df_country = (
    df_country
    .reset_index()
    .rename(columns={'index': 'Country Code'})
)

df_country.info()
df_country.head(2)
```

```python
#Reading file csv with PIB information

pib = pd.read_csv('../data/raw/gdp_data.csv')
pib.rename(columns={'Country Name': 'name'}, inplace=True)
pib.info()
pib.head(2)
```

```python
#Manipulation DataFrame

final_csv = pd.merge(df_country, pib, on=['Country Code', 'name'], how='inner')
final_csv.drop(columns=['currency', 'Indicator Name', 'Indicator Code', 'Unnamed: 66'], inplace=True)
final_csv.rename(columns={'Country Code': 'codigo'}, inplace=True)
final_csv.head()

```

```python
#Write final data in file csv

final_csv.to_csv('../data/processed/countries_gdp.csv', index=False)
```
